```python
import os
import shutil

# Thư mục cache của Hugging Face
HF_CACHE_DIR = os.path.join(os.getenv("HF_HOME", os.path.expanduser("~/.cache/huggingface/hub")))

MODELS_DIR = [d for d in os.listdir(HF_CACHE_DIR) if d.startswith("models--")]

def get_size(start_path='.'):
    """Tính dung lượng thư mục"""
    total_size = 0
    for dirpath, dirnames, filenames in os.walk(start_path):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if os.path.exists(fp):
                total_size += os.path.getsize(fp)
    return total_size

if not MODELS_DIR:
    print("Không tìm thấy model nào trong cache.")
else:
    print(f"Tìm thấy {len(MODELS_DIR)} model trong cache:")
    for idx, model in enumerate(MODELS_DIR, 1):
        path = os.path.join(HF_CACHE_DIR, model)
        size = get_size(path) / (1024**3)  # chuyển sang GB
        print(f"{idx}. {model} - {size:.2f} GB")

    # Hỏi có muốn xóa model không
    delete = input("\nBạn có muốn xóa một model để giải phóng dung lượng không? (y/n) ").lower()
    if delete == 'y':
        for idx, model in enumerate(MODELS_DIR, 1):
            print(f"{idx}. {model}")
        choice = input("Nhập số model muốn xóa: ")
        try:
            choice_idx = int(choice) - 1
            if 0 <= choice_idx < len(MODELS_DIR):
                path_to_delete = os.path.join(HF_CACHE_DIR, MODELS_DIR[choice_idx])
                shutil.rmtree(path_to_delete)
                print(f"Đã xóa {MODELS_DIR[choice_idx]}")
            else:
                print("Chọn không hợp lệ.")
        except Exception as e:
            print(f"Lỗi: {e}")
```
